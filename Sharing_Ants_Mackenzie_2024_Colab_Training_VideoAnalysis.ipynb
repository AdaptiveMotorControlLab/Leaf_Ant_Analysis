## Import Dependencies


```
#no need to change anything (unless you move the project):
import matplotlib
import os
import deeplabcut
from base64 import b64encode
from IPython.display import HTML

#/content/drive/MyDrive/Ants-mackenzie-2024-02-11/config.yaml
project_path = '/content/drive/My Drive/'+ProjectFolderName
config_path = os.path.join(project_path, "config.yaml")
```

## Train a model

```
deeplabcut.create_training_dataset(config_path)

#NOTE: ran this cell, then edited the dlc-models --> train pose_cfg.yaml
#using the defaults aside:
# from I changed the TrainFraction in config to .99 to train on most all the data
# sharpen: false --> true
# global_scale: 0.8 --> 0.9
# pos_dist_thresh: 17 --> 11
```

```
deeplabcut.train_network(config_path, shuffle=1, displayiters=100,saveiters=5000)
```

### Evaluate:
- note I am using 99% to train given the limited labeling in round 1, thus this # is not reliable; check video quality...


```
%matplotlib notebook
deeplabcut.evaluate_network(config_path,plotting=True)
```

## Export the model for sharing:
- note, it is known that this model is less confidient when the ant is in the leaf chamber; given the goal was to compute relative time in or out of the leaf area, we use the tracks for quantitative analysis for the open area.
- this model will not generalize to other ants or videos; it is highly domain specific.

```
deeplabcut.export_model(config_path)
```

## Start Analyzing a video with our trained model!
This function analyzes your video(s).

The results are stored in hd5 file in the same directory where the video resides. The data is optionally also exported in comma-separated values format (.csv).

```
#You can change this to be an uploaded video in Colab. On the left upload or drag&drop!

deeplabcut.analyze_videos(config_path,videofile_path, shuffle=1,videotype="mp4",auto_track=True, cropping=[300, 1440, 45, 840] )
```

```
deeplabcut.convert_detections2tracklets(config_path,videofile_path, shuffle=1, identity_only=True, overwrite = True)
```

```
deeplabcut.stitch_tracklets(config_path,videofile_path, shuffle=1)
```

You can also now filter your data to smooth any small jitters:

```
deeplabcut.filterpredictions(config_path,videofile_path, shuffle=1,videotype="mp4")
```

## Plot the trajectories of the analyzed videos:
This function plots the trajectories of all the body parts across the entire video. Each body part is identified by a unique color.

```
deeplabcut.plot_trajectories(config_path,[videofile_path], shuffle=1,videotype="mp4")
```

Now you can look at the plot-poses file and check the "plot-likelihood.png" might want to change the "p-cutoff" in the config.yaml file so that you have only high confidnece points plotted in the video. i.e. ~0.8 or 0.9. The current default is 0.4.
